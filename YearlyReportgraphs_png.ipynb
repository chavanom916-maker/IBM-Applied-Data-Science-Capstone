```python
import pandas as pd

URL = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/historical_automobile_sales.csv"
df = pd.read_csv(URL)
print("Dataset loaded!")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Choose a specific year, for example 2013
chosen_year = 2013
yearly_data = df[df['Year'] == chosen_year]

# Create the 4 yearly report graphs in a 2x2 grid layout
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Graph 1: Yearly Average Automobile Sales Trend
yas = df.groupby('Year')['Automobile_Sales'].mean().reset_index()
axes[0, 0].plot(yas['Year'], yas['Automobile_Sales'], marker='o', color='b')
axes[0, 0].set_title('Yearly Average Automobile Sales Trend')
axes[0, 0].set_xlabel('Year')
axes[0, 0].set_ylabel('Automobile Sales')

# Graph 2: Total Monthly Automobile Sales for the selected year
mas = yearly_data.groupby('Month')['Automobile_Sales'].sum().reset_index()
axes[0, 1].plot(mas['Month'], mas['Sales'] if 'Sales' in mas else mas['Automobile_Sales'], marker='o', color='g')
axes[0, 1].set_title(f'Total Monthly Automobile Sales for {chosen_year}')
axes[0, 1].set_xlabel('Month')
axes[0, 1].set_ylabel('Total Sales')

# Graph 3: Average Vehicles Sold by Vehicle Type in the selected year
avr_v = yearly_data.groupby('Vehicle_Type')['Automobile_Sales'].mean().reset_index()
axes[1, 0].bar(avr_v['Vehicle_Type'], avr_v['Automobile_Sales'], color='purple')
axes[1, 0].set_title(f'Average Vehicles Sold by Type in {chosen_year}')
axes[1, 0].tick_params(axis='x', rotation=45)

# Graph 4: Total Advertisement Expenditure by Vehicle Type in the selected year
exp_v = yearly_data.groupby('Vehicle_Type')['Advertising_Expenditure'].sum()
axes[1, 1].pie(exp_v, labels=exp_v.index, autopct='%1.1f%%', startangle=90)
axes[1, 1].set_title(f'Total Advertisement Expenditure by Vehicle Type ({chosen_year})')

plt.tight_layout()

# Save the figure as the exact file name required for Question 8
plt.savefig('YearlyReportgraphs.png')
print("YearlyReportgraphs.png created successfully!")
plt.show()
```
