```python
import pandas as pd

# Corrected stable URL with proper folder encoding
URL = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/historical_automobile_sales.csv"

df = pd.read_csv(URL)
print("Dataset loaded successfully!")
print(df.head())
```

```python
# Task 1.1
df_line = df.groupby('Year')['Automobile_Sales'].mean().reset_index()

plt.figure(figsize=(10, 5))
plt.plot(df_line['Year'], df_line['Automobile_Sales'], marker='o', color='b')
plt.title('Automobile Sales Fluctuation Over Time')
plt.xlabel('Year')
plt.ylabel('Average Automobile Sales')
plt.grid(True)
plt.show()
```

```python
# Task 1.2
non_rec = df[df['Recession'] == 0]
plt.figure(figsize=(10, 5))
sns.lineplot(data=non_rec, x='Year', y='Advertising_Expenditure', color='purple')
plt.title('Advertising Expenditure Trend in Non-Recession Periods')
plt.xlabel('Year')
plt.ylabel('Advertising Expenditure')
plt.grid(True)
plt.show()
```

```python
# Task 1.3 (Matches Question 2 answer)
plt.figure(figsize=(10, 5))
sns.barplot(data=df, x='Recession', y='Automobile_Sales', hue='Vehicle_Type')
plt.title('Vehicle Sales by Type: Recession vs Non-Recession')
plt.xlabel('Recession (0 = No, 1 = Yes)')
plt.ylabel('Average Automobile Sales')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

```python
# Task 1.4
fig, ax = plt.subplots(1, 2, figsize=(14, 5))

sns.lineplot(data=df[df['Recession'] == 1], x='Year', y='GDP', ax=ax[0], color='red')
ax[0].set_title('GDP Trend During Recession')

sns.lineplot(data=df[df['Recession'] == 0], x='Year', y='GDP', ax=ax[1], color='green')
ax[1].set_title('GDP Trend During Non-Recession')

plt.show()
```

```python
# Task 1.5
plt.figure(figsize=(10, 5))
sns.scatterplot(data=df, x='Seasonality_Weight', y='Automobile_Sales', size='Consumer_Confidence', hue='Vehicle_Type', sizes=(20, 200))
plt.title('Impact of Seasonality on Automobile Sales')
plt.show()
```

```python
# Task 1.6
rec_data = df[df['Recession'] == 1]
plt.figure(figsize=(10, 5))
sns.scatterplot(data=rec_data, x='Price', y='Automobile_Sales', hue='Vehicle_Type')
plt.title('Vehicle Price vs Sales Volume During Recessions')
plt.show()
```

```python
# Task 1.7 & 1.8
fig, ax = plt.subplots(1, 2, figsize=(14, 6))

adv_rec = df.groupby('Recession')['Advertising_Expenditure'].sum()
ax[0].pie(adv_rec, labels=['Non-Recession', 'Recession'], autopct='%1.1f%%', startangle=90)
ax[0].set_title('Advertising Expenditure: Recession vs Non-Recession')

adv_veh = rec_data.groupby('Vehicle_Type')['Advertising_Expenditure'].sum()
ax[1].pie(adv_veh, labels=adv_veh.index, autopct='%1.1f%%', startangle=90)
ax[1].set_title('Ad Expenditure by Vehicle Type During Recession')

plt.show()
```

```python
# Task 1.9
plt.figure(figsize=(10, 5))
sns.lineplot(data=rec_data, x='unemployment_rate', y='Automobile_Sales', hue='Vehicle_Type')
plt.title('Effect of Unemployment Rate on Sales during Recession')
plt.show()
```
